# Customer Dataset Cleaning

**loading the data_sets**

```python
customer_df =pd.read_csv('/Users/mac/DEV/Capstone_Project/pharmacy-customer-sales-intelligence/data/raw/Raw_data/customers.csv')
```

**Basic check pipeline**

```python
def basic_check(df):
    print(50*'=','DataSet Shape',50*'=')
    print(df.shape)
    print(50*'=','Variables Dtypes',50*'=')
    print(df.info())
    print(50*'=','Checking for dataset duplication',50*'=')
    print(df.duplicated().sum())
    print(50*'=','checking for dataset missingness',50*'=')
    print(df.isna().sum())
    print(50*'=','checking for dataset percentage missingness',50*'=')
    print((df.isna().sum()/len(df))*100)
    print('=' * 50, 'Statistical Summary', '=' * 50)
    display(df.describe().T)
    print('=' * 50, 'Categorical Summary', '=' * 50)
    display(df.describe(include=['object', 'category']).T)
```

```python
customer_df.head(5)
```

```python
basic_check(customer_df)
```

**Converting date variable to correct date format**

```python
customer_df['registration_date']=pd.to_datetime(customer_df['registration_date'],format='mixed',errors='coerce')
```

**Categorical columms entries standardization**

```python
cust_cat_cols = customer_df.select_dtypes(include='object').columns.to_list()
```

```python
cust_cat_cols
```

**checking unique entries**

```python
for col in cust_cat_cols[1:]:
    display(customer_df[col].unique())
```

**saving the cleaned customer dataset**

```python
customer_df.to_pickle('customer_clean.pkl')
```

**Converting_Date variable to correct date format**

```python
invent_df['movement_datetime']=pd.to_datetime(invent_df['movement_datetime'],format='mixed',errors='coerce')
```

**Dealing with the missingness**

```python
#droping the adjustment_reason column it has 98% missing entries
invent_df.drop(columns=['adjustment_reason'],inplace=True)
```

**Checking for unique values in categorical columns and standardizing the entries**

```python
invent_cat_cols = invent_df.select_dtypes(include='str').columns.tolist()
invent_cat_cols
```
